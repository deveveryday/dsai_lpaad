```python
times = [32, 48, 27, 55, 41, 29]
delivery_index = []
successful_delivery = []
failed_delivery = []
count = 0

for time in times:
  count += 1
  if time <= 40:
    successful_delivery.append(time)
  else:
    failed_delivery.append(time)

  delivery_index.append(["Delivery " + str(count), time, "sucessful" if time<= 40 else "fail"])
else:
  print("finish")


total = sum(times)
mean = total / len(times)
failed = len(failed_delivery)

percentual_success = (100 * len(successful_delivery)) / len(times)
percentual_fail = (100 * len(failed_delivery)) / len(times)
percentual_failed_delivery =  (1 - percentual_success) * 100

print("Index", delivery_index)
print("Total time", total, "minutes")
print("Mean time", f"{mean:.2f}", "minutes")
print("Fail count", failed, "deliveries")
print("Fail percentual", f"{percentual_fail:.2f}%")
```
